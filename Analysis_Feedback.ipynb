### Cleaning the Review Text

```python
def clean_text(text):
    """
    Clean customer review text by:
    1. Converting text to lowercase
    2. Removing special characters
    3. Removing extra whitespace
    """
    
    text = str(text).lower()
    text = re.sub(r"[^a-z0-9\s]", "", text)
    text = re.sub(r"\s+", " ", text).strip()
    
    return text
```

```python
# Clean the review text
df["Cleaned_Review"] = df["Review_Text"].apply(clean_text)

df[["Review_Text", "Cleaned_Review"]].head()
```

### Critical Reviews

Critical reviews are defined with a simple rule: a rating of 1 or 2 stars (no machine learning).

```python
# Critical reviews are defined as reviews with a rating of 1 or 2 stars
critical_reviews = df[df["Rating"] <= 2].copy()

print("Number of critical reviews:", len(critical_reviews))

critical_reviews
```

```python
from collections import Counter

# Combine all critical review text
critical_text = " ".join(critical_reviews["Cleaned_Review"])

# Split the text into individual words
words = critical_text.split()

# Words that signal a complaint (matched against the cleaned, lowercase review text)
complaint_keywords = {
    "broken", "damaged", "defective", "late", "delay", "delayed", "refund",
    "missing", "wrong", "rude", "poor", "bad", "unusable", "stopped",
    "failed", "issue", "problem", "replacement"
}

def get_top_complaints(reviews_df, keywords, top_n=None):
    """Count complaint keywords in the Cleaned_Review column and return them,
    most frequent first, as a DataFrame (top_n=None returns all matches)."""
    all_words = " ".join(reviews_df["Cleaned_Review"]).split()
    counts = Counter(word for word in all_words if word in keywords)
    return pd.DataFrame(counts.most_common(top_n),
                        columns=["Complaint Keyword", "Frequency"])
```

```python
# Run the function on the critical reviews
complaint_summary = get_top_complaints(critical_reviews, complaint_keywords)
complaint_summary
```

### Insight: Top Complaint Keywords

Across the 65 critical reviews, the most frequent complaint keywords are **damaged** (11), **replacement** (10) and **stopped** (9). Together with **defective** (5) and **broken** (3), these point mainly to product quality and durability problems: items arrive damaged or stop working.

**Late** (7) points to delivery delays, **missing** (6) suggests incomplete orders, and **rude** (5) and **refund** (2) point to customer service and billing issues.

### Selecting the 3 Most Critical, Detailed Reviews

Selection rule (applied to the filtered `critical_reviews` data):
1. Keep only the most critical reviews (1 star).
2. Remove duplicate review texts so the same complaint is not picked twice.
3. Take the 3 longest reviews, since longer text gives the AI more detail to respond to.

```python
# Select the 3 most critical (1-star) and most detailed (longest) reviews
top3_critical = (
    critical_reviews[critical_reviews["Rating"] == 1]
    .assign(Review_Length=lambda d: d["Review_Text"].str.len())
    .drop_duplicates(subset="Review_Text")
    .nlargest(3, "Review_Length")
)

top3_critical[["Review_ID", "Product", "Rating", "Customer_Name", "Review_Text"]]
```

```python
import os
import time
from dotenv import load_dotenv
from google import genai
from google.genai.errors import APIError
from IPython.display import Markdown, display

# 1. Load the API key from the .env file (it should contain: GEMINI_API_KEY=your_key)
load_dotenv()
if "GEMINI_API_KEY" not in os.environ:
    raise EnvironmentError("GEMINI_API_KEY not found. Add it to your .env file and restart the kernel.")
print("Gemini API key loaded from the .env file.")

# 2. Initialise the Gemini client (it reads GEMINI_API_KEY automatically)
client = genai.Client()

# 3. We ask Gemini to draft the apology email, retrying if the server is overloaded (503)
def generate_apology_email(customer_name, product_name, review_text, max_retries=3, initial_delay=4):
    prompt = f"""
    You are a professional, empathetic Customer Support Agent.
    Write a short, personalized apology email addressing the specific complaints in the review below.

    Customer Name: {customer_name}
    Product Purchased: {product_name}
    Customer Review: "{review_text}"

    Guidelines:
    - Show empathy and acknowledge the specific problems the customer mentioned.
    - Keep it under 120 words.
    - Do not promise specific refunds, replacements or timelines; say the team will review the case and follow up.
    - Include a subject line and end with a sign-off from "Customer Care Team".
    """

    delay = initial_delay
    for attempt in range(max_retries):
        try:
            # Gemini Flash-Lite model: fast and low-cost for short text generation
            response = client.models.generate_content(
                model="gemini-3.5-flash-lite",
                contents=prompt
            )
            return response.text

        except APIError as e:
            if e.code == 503:  # server overloaded: wait, then retry with a longer delay
                print(f"Server overloaded (attempt {attempt + 1}/{max_retries}). Retrying in {delay}s...")
                time.sleep(delay)
                delay *= 2
            else:
                return f"API Error ({e.code}): {e.message}"
        except Exception as e:
            return f"Unexpected Error: {str(e)}"

    return "Error: server remained unavailable after the maximum number of retries."

# 4. Generate and display an email for each selected review
results = []
for _, row in top3_critical.iterrows():
    email = generate_apology_email(row["Customer_Name"], row["Product"], row["Review_Text"])
    results.append({"Review_ID": row["Review_ID"], "Product": row["Product"],
                    "Review_Text": row["Review_Text"], "Generated_Email": email})
    display(Markdown(
        f"#### {row['Review_ID']} | {row['Product']} (Rating: {row['Rating']})\n\n"
        f"**Customer review:** {row['Review_Text']}\n\n---\n\n{email}"
    ))
    time.sleep(1)  # short pause between API calls

# 5. Save the emails so they can be reviewed before anything is sent
pd.DataFrame(results).to_csv("generated_emails.csv", index=False)
```
